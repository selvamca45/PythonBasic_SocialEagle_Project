Arithmatic operators

```python
a=10
b=2
print("Addition",a+b)
print("Subtraction",a-b)
print("Multiplication",a*b) 
print("Division",a/b)
print("Modulus",a%b)
print("Exponentiation",a**b)
print("Floor Division",a//b)
```

Comparison Operator

```python
p=10
q=20
print("comparison",p<q)
print("comparison",p>q)
print("comparison",p<=q)
print("comparison",p>=q)
print("comparison",p==q)
print("comparison",p!=q)
print("comparison",p is q)
print("comparison",p is not q)
print("comparison",p>q and q<=p)
print("comparison",p==q or q)
print("comparison",not p)
print("comparison",not q)
print("comparison",p & q)
print("comparison",p | q)  
print("comparison",p ^ q)
print("comparison",p << q)
print("comparison",p >> q)
print("comparison",p // q)
print("comparison",p % q)
print("comparison",p ** q)
```

Collections 

```python
fruits = ["apple", "banana", "cherry"]
print("fruits is",fruits)
print("first fruit is",fruits[0])
print("last fruit is",fruits[-1])  

fruits.append("orange")
print("fruits is",fruits)   

fruits.insert(1, "kiwi")
print("fruits is",fruits)

fruits.remove("banana")
print("fruits is",fruits)

colors = ("red", "green", "blue")
print("colors is",colors)
print("first color is",colors[0])
print("last color is",colors[-1])

numbers = {1, 2, 3, 4, 5,2,3}
print("numbers is",numbers)

student = {"name": "John", "age": 20, "city": "New York"}
print("student is",student)
print("student name is",student["name"])
print("student age is",student["age"])
print("student city is",student["city"])
student["city"] = "Los Angeles"
student["grade"] = "A"  
print("student is",student)
```
